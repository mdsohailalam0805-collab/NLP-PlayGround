# Now we apply Text Cleaning

```python
# Lowercasing the text

df["text"]=df["text"].str.lower()
df
```

```python
# removing the Punctuation

import string  #built in library
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))
df["text"]=df["text"].apply(remove_punctuation)
df
```

```python
# Remove the numbers from tha text

import re      #built in library (regular expressions[regex])
def remove_numbers(text):
    return re.sub(r"\d+","",text)       #sub is a functiion of regex which replaces the pattern with the given string
df["text"]=df["text"].apply(remove_numbers)


# Method2

def remove_numbers(text):
    new=""
    for ch in text:
        if ch.isdigit():
            continue
        else:
            new+=ch
    return new
df["text"]=df["text"].apply(remove_numbers)
df
```

```python
# Remove the urls and links from the text

from urlextract import URLExtract

extractor=URLExtract()
def remove_urls(text):
    urls=extractor.find_urls(text)
    for url in urls:
        text=text.replace(url,"")
    return text
df["text"]=df["text"].apply(remove_urls)
df  
```

```python
# Remove the html tags from the text

from bs4 import BeautifulSoup

def remove_html(text):
    return BeautifulSoup(text,"html.parser").get_text()
df["text"]=df["text"].apply(remove_html)
df
```

```python
# Remove the emojis from the text

def remove_emojis(text):
    new=""
    for ch in text:
        if ch.isascii():
            new+=ch
        else:
            continue
    return new
df["text"]=df["text"].apply(remove_emojis)
df
```

```python
# Rwmove the stopwords from the text

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("stopwords")
nltk.download("punkt_tab")
```

```python
stop_words=set(stopwords.words("english"))
stop_words
len(stop_words)
```

```python
def remove_stopwords(text):
    words = word_tokenize(text)
    new_words = []

    for word in words:
        if word not in stop_words:
            new_words.append(word)

    return " ".join(new_words)

df["text"]=df["text"].apply(remove_stopwords)
    
df.loc[1,"text"]
```
